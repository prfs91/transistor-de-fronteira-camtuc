## 1. Contextualização e Propósito Metodológico

Esta atividade compreende a construção de uma representação gráfica rigorosa para o monitoramento e a interpretação da evolução temporal das intenções de voto referentes a dois candidatos (**Candidato A** e **Candidato B**), com base em um histórico de sete pesquisas consecutivas. 

Na ciência de dados e na estatística, a transformação de matrizes numéricas em vetores visuais de linha permite a identificação imediata de tendências latentes, tais como vetores de crescimento, pontos de inflexão, períodos de estabilidade ou convergência estatística. A finalidade é traduzir o comportamento amostral em um fluxo contínuo de dados que revele a dinâmica macroscópica da disputa. Para extrair os padrões e garantir o rigor analítico da base, a exploração foi estruturada em duas frentes metodológicas complementares:

*   **Análise Espacial de Trajetórias (Line Plots):** Mapeamento do fluxo de intenção de votos ao longo de um eixo cronológico composto por sete medições, permitindo mensurar a volatilidade e as taxas de aceleração ou desaceleração de cada candidato.
*   **Mensuração de Intervalo de Confiança (Hastes de Erro Vertical):** Inclusão de um parâmetro de flutuação fixo para delimitar a variabilidade amostral aceitável e verificar, matematicamente, cenários de distanciamento conclusivo ou intersecções numéricas.

## 2. Estruturação dos Vetores e Preparação do Ambiente

O estágio inicial consiste na definição das variáveis e parâmetros que alimentam o modelo. Os dados tabulares fornecidos são estruturados em listas indexadas e a margem de erro amostral é formalizada no ambiente de execução.

| Amostra Temporal | Pesquisa 1 | Pesquisa 2 | Pesquisa 3 | Pesquisa 4 | Pesquisa 5 | Pesquisa 6 | Pesquisa 7 |
| :--- | :---: | :---: | :---: | :---: | :---: | :---: | :---: |
| **Candidato A (%)** | 30 | 32 | 40 | 37 | 38 | 50 | 48 |
| **Candidato B (%)** | 22 | 25 | 30 | 31 | 33 | 37 | 40 |

```python
import matplotlib.pyplot as plt

# Séries numéricas fornecidas representando o histórico de intenção de votos
candidato_a = [30, 32, 40, 37, 38, 50, 48]
candidato_b = [22, 25, 30, 31, 33, 37, 40]

# Geração automática da sequência cronológica das pesquisas (de 1 a 7)
pesquisas = [str(i) for i in range(1, 8)]

# Definição do intervalo de confiança (margem de erro fixa de 1 ponto percentual)
margem_erro = 1
```

---

## 3. Construção e Renderização Integrada do Gráfico

Para conferir o rigor estatístico exigido, o algoritmo utiliza o método `plt.errorbar` em detrimento de uma plotagem linear simples. Esta abordagem introduz hastes verticais simétricas sobre cada coordenada, delimitando visualmente o intervalo real de flutuação dos dados.

### 📐 Anatomia dos Parâmetros Utilizados:
> * `figsize=(10, 6)`: Define uma proporção retangular áurea, otimizando o rastreamento ocular e a leitura em telas digitais.
> * `fmt='-o'`: Sintaxe que dita a identidade visual da linha (traço contínuo acompanhado de marcadores circulares nos pontos de dados).
> * `capsize=4`: Extensões horizontais nas extremidades das barras de erro, conferindo acabamento estético e facilitando a leitura dos limites superiores e inferiores.
> * `plt.grid()`: Malha de coordenadas em segundo plano com opacidade reduzida, servindo como guia visual analítico sem poluir o primeiro plano.

<div class="alert alert-block alert-warning" style="border-left: 5px solid #ff9800; padding: 12px; background-color: #fff9e6; border-radius: 4px;">
    <strong>⚠️ Nota de Execução do Jupyter:</strong> Toda a lógica de construção e formatação textual do gráfico deve ser concentrada estritamente dentro da mesma célula de código. Caso contrário, o interpretador gerará fragmentos visuais isolados ou telas nulas.
</div>

```python
# 1. Inicialização e dimensionamento do espaço gráfico
plt.figure(figsize=(10, 6))

# 2. Mapeamento e plotagem da trajetória do Candidato A (Linha Azul)
plt.errorbar(
    pesquisas,
    candidato_a,
    yerr=margem_erro,
    fmt="-o",
    capsize=4,
    color="#1f77b4",
    linewidth=2,
    markersize=6,
    label="Candidato A",
)

# 3. Mapeamento e plotagem da trajetória do Candidato B (Linha Vermelha)
plt.errorbar(
    pesquisas,
    candidato_b,
    yerr=margem_erro,
    fmt="-o",
    capsize=4,
    color="#d62728",
    linewidth=2,
    markersize=6,
    label="Candidato B",
)

# 4. Atribuição das nomenclaturas formais dos eixos coordenados
plt.xlabel("Pesquisa", fontsize=11, fontweight="bold", labelpad=10)
plt.ylabel("Porcentagem de Votos (%)", fontsize=11, fontweight="bold", labelpad=10)

# 5. Inserção de metadados informativos (Título descritivo e Legenda de identificação)
plt.title(
    "Evolução Temporal das Intenções de Voto com Margem de Erro (1%)",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
plt.legend(loc="upper left", fontsize=10, frameon=True, shadow=True)

# 6. Ativação da grade de referência em segundo plano
plt.grid(True, linestyle="--", alpha=0.5)

# 7. Finalização e exibição do gráfico consolidado
plt.show()
```
