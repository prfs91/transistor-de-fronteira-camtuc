### 9. Regressão Linear

A Regressão Linear busca modelar relações lineares entre variáveis independentes e dependentes.

```python
modelo_linear, tempo_linear = treinar_regressao_linear(
    X_treino,
    y_treino
)

y_pred_linear = realizar_predicoes(
    modelo_linear,
    X_teste
)

metricas_linear = calcular_metricas(
    y_teste,
    y_pred_linear
)

metricas_linear
```

### 10. Support Vector Regression (SVR)

O SVR utiliza hiperplanos para realizar regressão e é sensível à escala dos dados, sendo importante utilizar padronização.

```python
modelo_svr, tempo_svr = treinar_svr(
    X_treino,
    y_treino
)

y_pred_svr = realizar_predicoes(
    modelo_svr,
    X_teste
)

metricas_svr = calcular_metricas(
    y_teste,
    y_pred_svr
)

metricas_svr
```

### 11. Gaussian Process Regression (GPR)

O GPR é um modelo probabilístico capaz de estimar incertezas nas predições.

```python
modelo_gpr, tempo_gpr = treinar_gpr(
    X_treino,
    y_treino
)

y_pred_gpr, desvio_padrao = realizar_predicoes_gpr(
    modelo_gpr,
    X_teste
)

metricas_gpr = calcular_metricas(
    y_teste,
    y_pred_gpr
)

metricas_gpr
```

### 12. Validação Cruzada

```python
scores_linear = executar_validacao_cruzada(
    modelo_linear,
    X,
    y
)

scores_svr = executar_validacao_cruzada(
    modelo_svr,
    X,
    y
)

scores_gpr = executar_validacao_cruzada(
    modelo_gpr,
    X,
    y
)

print("Linear:", scores_linear.mean())
print("SVR:", scores_svr.mean())
print("GPR:", scores_gpr.mean())
```

Enquanto a Regressão Linear apresenta estrutura mais simples e poucos parâmetros configuráveis, modelos como SVR e GPR exigem ajuste de hiperparâmetros relacionados à flexibilidade do modelo e comportamento do kernel.
